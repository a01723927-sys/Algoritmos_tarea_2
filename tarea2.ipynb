### Pregunta 3 

Utilizando algoritmos con la variable mis_datos, responde las siguientes preguntas: 

¿Qué marca produce más vehículos y qué marca produce menos? 

```python
#make es marca
items = mis_datos['Make'].value_counts()
maxproductionid = items.idxmax()
maxproduction = items.max()

minproductionid = items.idxmin()
minproduction = items.min()

print(f"La marca con mayor producción de vehículos es {maxproductionid} con {maxproduction} vehículos.")
print(f"La marca con menor producción de vehículos es {minproductionid} con {minproduction} vehículos.")
```

### Pregunta 8 

Con la información de la variable mis_nuevos_datos, crea una nueva variable llamada opinion que esté en función del precio sugerido para la venta. Si el precio es mayor a 60,000 el valor de la variable opinion debe ser "Expensive", si el precio está entre 60,000 y 30,000, la variable opinion debe decir "Normal". Y si el precio es menor a 30,000 la variable opinion debe sugerir "Inexpensive". Almacena la información en la variable mis_nuevos_datos 

```python
# price > 60000 = expensive
# price > 30000 y < 60000 = Normal
# price < 30000 = Inexpensive

def pricetag(MSRP):
    if MSRP > 60000:
        return "expensive"
    elif MSRP < 30000:
        return "Inexpensive"
    else:
        return "Normal"

mis_nuevos_datos["opinion"] = (mis_nuevos_datos["MSRP"].apply(pricetag))

mis_nuevos_datos.head()
```
